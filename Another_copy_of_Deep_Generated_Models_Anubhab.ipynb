```python
# GAN for Handwriting Synthesis (Conditional GAN on MNIST)
# Full end-to-end code: data, models, training loop, checkpoints, visuals & analysis.

# ------------------------
# Step 0: Install & imports
# ------------------------
!pip install tensorflow==2.12.0 --quiet

import os
import numpy as np
import matplotlib.pyplot as plt
from tqdm import trange
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers
from tensorflow.keras.utils import to_categorical

print("TensorFlow version:", tf.__version__)

# ------------------------
# Step 1: Settings & folders
# ------------------------
OUTPUT_DIR = "/content/gan_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(os.path.join(OUTPUT_DIR, "samples"), exist_ok=True)
os.makedirs(os.path.join(OUTPUT_DIR, "models"), exist_ok=True)

# Training hyperparameters (tune these)
BUFFER_SIZE = 60000
BATCH_SIZE = 256
NOISE_DIM = 100
NUM_CLASSES = 10
EMBED_DIM = 50         # label embedding dimension
EPOCHS = 50            # adjust if you want longer training
SAVE_EVERY = 5         # save samples every N epochs
GEN_LR = 2e-4
DISC_LR = 2e-4
BETA_1 = 0.5           # recommended for GANs

# ------------------------
# Step 2: Load & preprocess MNIST
# ------------------------
(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.mnist.load_data()
# Combine train+test optionally (we'll use train only)
images = train_images.astype('float32')
labels = train_labels.astype('int32')

# Normalize to [-1, 1]
images = (images - 127.5) / 127.5
# Expand channel dimension
images = np.expand_dims(images, axis=-1)  # shape (N,28,28,1)

print("Images shape:", images.shape)
print("Labels shape:", labels.shape)

# Create tf.data dataset
dataset = tf.data.Dataset.from_tensor_slices((images, labels))
dataset = dataset.shuffle(BUFFER_SIZE).batch(BATCH_SIZE, drop_remainder=True).prefetch(tf.data.AUTOTUNE)

# ------------------------
# Step 3: Conditional Generator & Discriminator
# ------------------------
# Approach: embed label, concatenate with noise for generator.
# For discriminator: embed label and concatenate as extra channels to image.

def build_generator(noise_dim=NOISE_DIM, embed_dim=EMBED_DIM):
    noise_input = layers.Input(shape=(noise_dim,), name="noise_input")
    label_input = layers.Input(shape=(1,), dtype='int32', name="label_input")
    # label embedding -> dense -> reshape -> concat with noise
    label_embedding = layers.Embedding(NUM_CLASSES, embed_dim, input_length=1)(label_input)
    label_embedding = layers.Flatten()(label_embedding)  # shape (embed_dim,)
    x = layers.Concatenate()([noise_input, label_embedding])  # combined vector

    # project and reshape
    x = layers.Dense(7*7*128, use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Reshape((7,7,128))(x)

    x = layers.Conv2DTranspose(64, kernel_size=5, strides=2, padding='same', use_bias=False)(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU(0.2)(x)

    x = layers.Conv2DTranspose(1, kernel_size=5, strides=2, padding='same', activation='tanh')(x)  # [-1,1]
    model = models.Model(inputs=[noise_input, label_input], outputs=x, name="Generator")
    return model

def build_discriminator(embed_dim=EMBED_DIM):
    img_input = layers.Input(shape=(28,28,1), name="img_input")
    label_input = layers.Input(shape=(1,), dtype='int32', name="label_input")
    # label embedding -> dense -> reshape to (28,28,1) to concat with image
    label_embedding = layers.Embedding(NUM_CLASSES, embed_dim, input_length=1)(label_input)
    label_embedding = layers.Dense(28*28, activation=None)(label_embedding)
    label_embedding = layers.Reshape((28,28,1))(label_embedding)

    # concatenate along channels
    x = layers.Concatenate(axis=-1)([img_input, label_embedding])  # shape (28,28,2)

    x = layers.Conv2D(64, kernel_size=5, strides=2, padding='same')(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Conv2D(128, kernel_size=5, strides=2, padding='same')(x)
    x = layers.LeakyReLU(0.2)(x)
    x = layers.Dropout(0.3)(x)

    x = layers.Flatten()(x)
    x = layers.Dense(1)(x)  # logits
    model = models.Model(inputs=[img_input, label_input], outputs=x, name="Discriminator")
    return model

generator = build_generator()
discriminator = build_discriminator()

generator.summary()
discriminator.summary()

# ------------------------
# Step 4: Losses & Optimizers
# ------------------------
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

# Discriminator loss: real_loss + fake_loss
def discriminator_loss(real_logits, fake_logits):
    real_loss = cross_entropy(tf.ones_like(real_logits)*0.9, real_logits)  # label smoothing: real=0.9
    fake_loss = cross_entropy(tf.zeros_like(fake_logits), fake_logits)
    return real_loss + fake_loss

# Generator loss: try to fool discriminator
def generator_loss(fake_logits):
    return cross_entropy(tf.ones_like(fake_logits), fake_logits)

gen_optimizer = optimizers.Adam(learning_rate=GEN_LR, beta_1=BETA_1)
disc_optimizer = optimizers.Adam(learning_rate=DISC_LR, beta_1=BETA_1)

# ------------------------
# Step 5: Training utilities
# ------------------------
@tf.function
def train_step(real_images, real_labels):
    # Sample noise & labels for generator
    batch_size = tf.shape(real_images)[0]
    noise = tf.random.normal([batch_size, NOISE_DIM])
    random_labels = tf.random.uniform([batch_size,1], minval=0, maxval=NUM_CLASSES, dtype=tf.int32)

    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        # generate images
        generated_images = generator([noise, random_labels], training=True)

        # discriminator outputs
        real_logits = discriminator([real_images, tf.expand_dims(real_labels,1)], training=True)
        fake_logits = discriminator([generated_images, random_labels], training=True)

        # losses
        gen_loss = generator_loss(fake_logits)
        disc_loss = discriminator_loss(real_logits, fake_logits)

    # gradients
    gradients_of_generator = gen_tape.gradient(gen_loss, generator.trainable_variables)
    gradients_of_discriminator = disc_tape.gradient(disc_loss, discriminator.trainable_variables)

    # apply
    gen_optimizer.apply_gradients(zip(gradients_of_generator, generator.trainable_variables))
    disc_optimizer.apply_gradients(zip(gradients_of_discriminator, discriminator.trainable_variables))

    return gen_loss, disc_loss

# fixed seed for sampling across epochs (for consistent visualization)
NUM_EXAMPLES_TO_GENERATE = 100
fixed_noise = tf.random.normal([NUM_EXAMPLES_TO_GENERATE, NOISE_DIM])
# prepare fixed label pattern: 10 labels repeated
fixed_labels = np.array([i for i in range(NUM_CLASSES) for _ in range(NUM_EXAMPLES_TO_GENERATE//NUM_CLASSES)])
fixed_labels = fixed_labels.reshape(-1,1)

# ------------------------
# Step 6: Training loop
# ------------------------
def generate_and_save_images(model, epoch, noise, labels, path):
    preds = model.predict([noise, labels], verbose=0)
    # rescale [-1,1] -> [0,1]
    imgs = (preds + 1.0) / 2.0
    # grid 10x10 for 100 samples
    fig = plt.figure(figsize=(10,10))
    for i in range(imgs.shape[0]):
        plt.subplot(10,10,i+1)
        plt.imshow(imgs[i,:,:,0], cmap='gray')
        plt.axis('off')
    plt.suptitle(f"Epoch {epoch}")
    plt.tight_layout()
    filename = os.path.join(path, f"sample_epoch_{epoch:03d}.png")
    plt.savefig(filename, dpi=150)
    plt.close(fig)

# training history
gen_losses = []
disc_losses = []

# Training loop
print("Starting training...")
for epoch in range(1, EPOCHS+1):
    epoch_gen_loss = []
    epoch_disc_loss = []

    for real_batch_images, real_batch_labels in dataset:
        g_loss, d_loss = train_step(real_batch_images, real_batch_labels)
        epoch_gen_loss.append(g_loss.numpy())
        epoch_disc_loss.append(d_loss.numpy())

    gen_mean = np.mean(epoch_gen_loss)
    disc_mean = np.mean(epoch_disc_loss)
    gen_losses.append(gen_mean)
    disc_losses.append(disc_mean)

    print(f"Epoch {epoch}/{EPOCHS}  Gen loss: {gen_mean:.4f}  Disc loss: {disc_mean:.4f}")

    # Save samples at checkpoints
    if epoch % SAVE_EVERY == 0 or epoch == 1 or epoch == EPOCHS:
        generate_and_save_images(generator, epoch, fixed_noise, fixed_labels, os.path.join(OUTPUT_DIR,"samples"))
        generator.save(os.path.join(OUTPUT_DIR,"models",f"generator_epoch_{epoch}.keras"))
        discriminator.save(os.path.join(OUTPUT_DIR,"models",f"discriminator_epoch_{epoch}.keras"))

# Save final models
generator.save(os.path.join(OUTPUT_DIR,"models","generator_final.keras"))
discriminator.save(os.path.join(OUTPUT_DIR,"models","discriminator_final.keras"))

# ------------------------
# Step 7: Plot losses & show sample images
# ------------------------
# Loss curves
plt.figure(figsize=(8,5))
plt.plot(range(1,len(gen_losses)+1), gen_losses, label='Generator loss')
plt.plot(range(1,len(disc_losses)+1), disc_losses, label='Discriminator loss')
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GAN Losses")
plt.legend()
plt.grid(True)
plt.savefig(os.path.join(OUTPUT_DIR, "loss_curves.png"), dpi=150)
plt.show()

# Show last checkpoint sample
last_sample = os.path.join(OUTPUT_DIR, "samples", f"sample_epoch_{EPOCHS:03d}.png")
if os.path.exists(last_sample):
    from IPython.display import Image, display
    print("Latest sample image:")
    display(Image(last_sample))
else:
    print("No sample image saved.")

# ------------------------
# Step 8: Mode collapse check (simple heuristics)
# ------------------------
# Load all sample images saved and compute variance across the grid as a crude check.
import glob
from PIL import Image as PILImage

sample_files = sorted(glob.glob(os.path.join(OUTPUT_DIR,"samples","sample_epoch_*.png")))
if len(sample_files) > 0:
    # examine last saved sample grid (the PNG)
    img = PILImage.open(sample_files[-1]).convert('L')
    arr = np.array(img).astype(np.float32) / 255.0
    variance = arr.var()
    print(f"Variance in pixel intensities for last sample grid: {variance:.6f}")
    if variance < 0.0005:
        print("Warning: very low variance — possible mode collapse (generator producing near-identical outputs).")
    else:
        print("Pixel variance OK — likely diverse outputs.")
else:
    print("No sample PNGs found for mode collapse heuristic.")

# ------------------------
# Step 9: Recommendations & printed analysis
# ------------------------
print("\n\n=== ANALYSIS & SUGGESTIONS ===")
print("What this code produced:")
print(f"- Trained a conditional GAN for {EPOCHS} epochs on MNIST (digits 0-9).")
print("- Saved generator/discriminator models in:", os.path.join(OUTPUT_DIR,"models"))
print("- Saved sample grids every", SAVE_EVERY, "epochs under:", os.path.join(OUTPUT_DIR,"samples"))
print("- Saved loss curves at:", os.path.join(OUTPUT_DIR,"loss_curves.png"))

print("\nHow to inspect training behaviour (what to look for):")
print("- Generator loss decreasing steadily while discriminator loss collapses to 0 may indicate unstable training.")
print("- If generated sample grids show identical or near-identical digits across many labels, that's MODE COLLAPSE.")

print("\nSuggested ways to stabilize training (if you observe issues):")
print("1) Label smoothing for discriminator (we used real=0.9) — helps prevent overconfidence.")
print("2) Use smaller learning rates and beta_1=0.5 (we used these settings).")
print("3) Use batch normalization in generator (we used it) and avoid it in discriminator's first layer.")
print("4) Add noise to discriminator inputs (e.g., Gaussian noise to real/fake images) to regularize.")
print("5) Use gradient penalty (WGAN-GP) or spectral normalization for discriminator to stabilize gradients.")
print("6) Reduce batch size or adjust model capacity if overfitting/underfitting occurs.")
print("7) Monitor diversity metrics (e.g., pixel variance, nearest neighbour distances) to detect mode collapse earlier.")
print("8) Train discriminator less aggressively (e.g., 1 discriminator step per generator step) or vice versa, experiment.")
print("\nMode collapse specific mitigation:")
print("- Minibatch discrimination, feature matching loss, unrolled GANs, WGAN variants.")

print("\nYou can view saved sample images in the 'samples' directory and loss curves 'loss_curves.png'.")
print("End of run.")
```
